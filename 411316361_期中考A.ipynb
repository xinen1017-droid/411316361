````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型（請勿修改）
# ============================================================
!pip install -q groq gradio

import os
import re
import time
import sqlite3
import pandas as pd
from IPython.display import display
from groq import Groq
import gradio as gr

# 判斷目前是在 Colab 還是本地端 Jupyter
try:
    from google.colab import userdata
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# 🔴 金鑰名稱一律為 GROQ_API_KEY；絕對不可以把金鑰寫在程式裡，也不可以 print 出來
if IN_COLAB:
    try:
        GROQ_API_KEY = userdata.get('GROQ_API_KEY')
    except Exception:
        print("❌ 讀不到金鑰：請到左側 🔑 新增名稱為 GROQ_API_KEY 的密鑰，並打開「筆記本存取權」")
        raise
else:
    GROQ_API_KEY = os.environ.get('GROQ_API_KEY')
    if not GROQ_API_KEY:
        raise RuntimeError("找不到環境變數 GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

# 期中考指定模型；推理強度設為 low，可大幅減少 token 用量
# （唯一允許修改處：若 20b 額度用完，可改為 "openai/gpt-oss-120b"，並在規格書最後註明）
MODEL = "openai/gpt-oss-20b"
REASONING_EFFORT = "low"
print("✅ 環境：", "Colab" if IN_COLAB else "本地端", "｜模型：", MODEL)
````

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# ============================================================
DB_NAME = "activity_registration.db"

def get_connection():
    """建立 SQLite 連線，並啟用外鍵功能。"""
    conn = sqlite3.connect(DB_NAME)
    conn.execute("PRAGMA foreign_keys = ON")
    return conn

conn = get_connection()
try:
    cur = conn.cursor()
    cur.execute("DROP TABLE IF EXISTS registrations")
    cur.execute("DROP TABLE IF EXISTS activities")

    cur.execute("""
        CREATE TABLE activities (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT NOT NULL UNIQUE,
            event_date TEXT NOT NULL,
            capacity INTEGER NOT NULL CHECK(capacity >= 0)
        )
    """)
    cur.execute("""
        CREATE TABLE registrations (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            student_name TEXT NOT NULL,
            activity_id INTEGER NOT NULL,
            people INTEGER NOT NULL CHECK(people BETWEEN 1 AND 3),
            status TEXT NOT NULL CHECK(status IN ('已報名', '已取消')),
            created_at TEXT NOT NULL DEFAULT CURRENT_TIMESTAMP,
            FOREIGN KEY(activity_id) REFERENCES activities(id)
        )
    """)

    initial_activities = [
        ("迎新晚會", "2026-11-20", 6),
        ("程式工作坊", "2026-11-25", 4),
        ("籃球友誼賽", "2026-11-28", 3)
    ]
    cur.executemany(
        "INSERT INTO activities (name, event_date, capacity) VALUES (?, ?, ?)",
        initial_activities
    )
    conn.commit()
    print("✅ 資料庫建立完成，已新增三場初始活動。")
finally:
    conn.close()
````

````python
# ============================================================
# 第 3 格：顯示資料表
# ============================================================
def show_tables():
    conn = get_connection()
    try:
        print("【活動資料表】")
        display(pd.read_sql_query("SELECT * FROM activities ORDER BY id", conn))
        print("【報名紀錄資料表】")
        display(pd.read_sql_query("SELECT * FROM registrations ORDER BY id", conn))
    finally:
        conn.close()

show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# ============================================================
ACTIVITIES = ["迎新晚會", "程式工作坊", "籃球友誼賽"]
ACTIONS = ["報名", "取消", "查詢", "統計"]
ai_cache = {}

def interpret_request(user_text):
    """呼叫 Groq 判斷動作與活動；程式再自行驗證允許值及人數。"""
    if user_text in ai_cache:
        return ai_cache[user_text]

    prompt = f"""
你是活動報名系統的文字分類器。請只根據使用者句子分類，不要執行任何操作。
使用者句子：{user_text}
動作只能選：報名、取消、查詢、統計、未知。
活動只能選：迎新晚會、程式工作坊、籃球友誼賽、全部、未知。
若詢問所有活動或報名統計，動作為統計、活動為全部。
若無法明確判斷，動作與活動都輸出未知。
人數欄位請照句意輸出數量；「我和兩個朋友」代表 3。
沒有提到人數時輸出 1。
請嚴格只輸出一行，格式：動作,活動,人數
不要輸出解釋、標點或 Markdown。
"""
    raw = ""
    for attempt in range(3):
        try:
            response = client.chat.completions.create(
                model=MODEL,
                messages=[{"role": "user", "content": prompt}],
                temperature=0,
                reasoning_effort=REASONING_EFFORT
            )
            raw = response.choices[0].message.content or ""
            time.sleep(2)
            break
        except Exception as e:
            error_text = str(e)
            if "429" in error_text and attempt < 2:
                print("⚠️ AI 使用量暫時超過限制，等待 30 秒後重試...")
                time.sleep(30)
            else:
                raw = "未知,未知,1"
                print("⚠️ AI 呼叫失敗，改以未知處理：", error_text[:180])
                break

    # 只從 AI 回覆中尋找允許的動作及活動，不直接相信整段回覆
    action = next((x for x in ACTIONS if x in raw), "未知")
    activity = "全部" if action == "統計" and "全部" in raw else next(
        (x for x in ACTIVITIES if x in raw), "未知"
    )
    # 人數以正規表示式擷取；找不到數字時預設 1
    match = re.search(r"\d+", raw)
    people = int(match.group()) if match else 1

    # 統計不需要特定活動；其他動作必須有可辨識活動
    if action == "統計":
        activity = "全部"
    if action not in ACTIONS:
        action, activity = "未知", "未知"

    result = (action, activity, people, raw)
    ai_cache[user_text] = result
    return result
````

````python
# ============================================================
# 第 5 格：AI 解讀測試（共 9 句）
# ============================================================
ai_tests = [
    ("我要報名迎新晚會", "報名", "迎新晚會", 1),
    ("幫我報名程式工作坊，我和兩個朋友一起去", "報名", "程式工作坊", 3),
    ("我要取消籃球友誼賽", "取消", "籃球友誼賽", 1),
    ("迎新晚會還有多少名額？", "查詢", "迎新晚會", 1),
    ("請列出各活動報名狀況", "統計", "全部", 1),
    ("幫我報名迎新晚會 3 人", "報名", "迎新晚會", 3),
    ("我要報名桌遊活動", "未知", "未知", 1),
    ("我要報名程式工作坊 4 人", "報名", "程式工作坊", 4),
    ("今天天氣如何", "未知", "未知", 1)
]

passed = 0
for i, (sentence, expected_action, expected_activity, expected_people) in enumerate(ai_tests, 1):
    action, activity, people, raw = interpret_request(sentence)
    ok = (
        (action == expected_action or expected_action == "*") and
        (activity == expected_activity or expected_activity == "*") and
        (people == expected_people or expected_people == "*")
    )
    passed += int(ok)
    print(f"測試 {i}：{'✅ 通過' if ok else '❌ 未通過'}")
    print("  使用者：", sentence)
    print("  AI 原始回覆：", raw)
    print("  系統解讀：", (action, activity, people))
    print("  預期結果：", (expected_action, expected_activity, expected_people))
print(f"\nAI 解讀測試：{passed}/{len(ai_tests)} 句通過")
````

````python
# ============================================================
# 第 6 格：功能函式
# ============================================================
def get_activity_summary(activity_name):
    """取得單一活動名額資訊。"""
    conn = get_connection()
    try:
        cur = conn.cursor()
        cur.execute("""
            SELECT a.id, a.capacity,
                   COALESCE(SUM(CASE WHEN r.status = '已報名' THEN r.people ELSE 0 END), 0)
            FROM activities AS a
            LEFT JOIN registrations AS r ON a.id = r.activity_id
            WHERE a.name = ?
            GROUP BY a.id, a.capacity
        """, (activity_name,))
        row = cur.fetchone()
        if row is None:
            return None
        activity_id, capacity, registered = row
        return activity_id, capacity, registered, capacity - registered
    finally:
        conn.close()

def register_student(student_name, activity_name, people):
    if not student_name or not student_name.strip():
        return "❌ 報名需要填寫姓名。"
    if people < 1 or people > 3:
        return "❌ 每次報名人數必須在 1～3 人之間。"

    conn = get_connection()
    try:
        cur = conn.cursor()
        cur.execute("SELECT id, capacity FROM activities WHERE name = ?", (activity_name,))
        activity = cur.fetchone()
        if activity is None:
            return "❌ 找不到這個活動。目前活動：" + "、".join(ACTIVITIES)
        activity_id, capacity = activity

        cur.execute("""
            SELECT id FROM registrations
            WHERE student_name = ? AND activity_id = ? AND status = '已報名'
        """, (student_name.strip(), activity_id))
        existing = cur.fetchone()
        if existing is not None:
            return "❌ 你已經報名這個活動，不可重複報名。"

        cur.execute("""
            SELECT COUNT(DISTINCT activity_id) FROM registrations
            WHERE student_name = ? AND status = '已報名'
        """, (student_name.strip(),))
        count_row = cur.fetchone()
        active_count = count_row[0] if count_row is not None else 0
        if active_count >= 2:
            return "❌ 每位學生最多同時報名 2 個不同活動。"

        cur.execute("""
            SELECT COALESCE(SUM(people), 0) FROM registrations
            WHERE activity_id = ? AND status = '已報名'
        """, (activity_id,))
        total_row = cur.fetchone()
        registered = total_row[0] if total_row is not None else 0
        remaining = capacity - registered
        if people > remaining:
            return f"❌ 名額不足！目前只剩 {remaining} 個名額。"

        cur.execute("""
            INSERT INTO registrations (student_name, activity_id, people, status)
            VALUES (?, ?, ?, '已報名')
        """, (student_name.strip(), activity_id, people))
        conn.commit()
        return f"✅ {student_name} 已成功報名「{activity_name}」，共 {people} 人。"
    finally:
        conn.close()

def cancel_registration(student_name, activity_name):
    if not student_name or not student_name.strip():
        return "❌ 取消需要填寫姓名。"
    conn = get_connection()
    try:
        cur = conn.cursor()
        cur.execute("""
            SELECT id FROM registrations
            WHERE student_name = ? AND activity_id = (
                SELECT id FROM activities WHERE name = ?
            ) AND status = '已報名'
            ORDER BY id DESC LIMIT 1
        """, (student_name.strip(), activity_name))
        row = cur.fetchone()
        if row is None:
            return f"❌ 查不到 {student_name} 的「{activity_name}」有效報名紀錄。"
        cur.execute("UPDATE registrations SET status = '已取消' WHERE id = ?", (row[0],))
        conn.commit()
        return f"✅ 已取消 {student_name} 的「{activity_name}」報名，名額已釋出。"
    finally:
        conn.close()

def query_remaining(activity_name):
    summary = get_activity_summary(activity_name)
    if summary is None:
        return "❌ 找不到這個活動。目前活動：" + "、".join(ACTIVITIES)
    _, capacity, registered, remaining = summary
    return f"「{activity_name}」總名額 {capacity} 人，已報名 {registered} 人，剩餘 {remaining} 人。"

def show_all_statistics():
    """LEFT JOIN 確保尚無報名紀錄的活動也會顯示。"""
    conn = get_connection()
    try:
        query = """
            SELECT a.name AS 活動, a.event_date AS 日期, a.capacity AS 名額,
                   COALESCE(SUM(CASE WHEN r.status = '已報名' THEN r.people ELSE 0 END), 0) AS 已報名人數,
                   a.capacity - COALESCE(SUM(CASE WHEN r.status = '已報名' THEN r.people ELSE 0 END), 0) AS 剩餘名額
            FROM activities AS a
            LEFT JOIN registrations AS r ON a.id = r.activity_id
            GROUP BY a.id, a.name, a.event_date, a.capacity
            ORDER BY a.id
        """
        df = pd.read_sql_query(query, conn)
        return df
    finally:
        conn.close()
````

````python
# ============================================================
# 第 7 格：主流程與流程測試（共 9 步）
# ============================================================
def handle_request(student_name, user_text):
    action, activity, people, raw = interpret_request(user_text)

    if action == "未知" or (action != "統計" and activity == "未知"):
        return "❓ 無法判斷您的需求。目前活動：" + "、".join(ACTIVITIES)
    if action in ["報名", "取消"] and (not student_name or not student_name.strip()):
        return "❌ 請先輸入姓名。"
    if action == "報名":
        return register_student(student_name, activity, people)
    if action == "取消":
        return cancel_registration(student_name, activity)
    if action == "查詢":
        return query_remaining(activity)
    if action == "統計":
        return show_all_statistics().to_string(index=False)
    return "❓ 無法處理此需求。"

# 每次流程測試前資料庫已由第 2 格重建，因此結果可重現
flow_tests = [
    ("王小明", "我要報名迎新晚會"),
    ("王小明", "我要報名迎新晚會"),
    ("陳小華", "我要報名程式工作坊，我和兩個朋友一起去"),
    ("林大同", "我要報名程式工作坊"),
    ("王小明", "我要報名籃球友誼賽"),
    ("王小明", "我要報名程式工作坊"),
    ("王小明", "取消迎新晚會"),
    ("趙小美", "迎新晚會還有多少名額？"),
    ("幹部", "請列出各活動報名狀況")
]

for step, (name, sentence) in enumerate(flow_tests, 1):
    print(f"\n【流程測試 {step}】{name}：{sentence}")
    print(handle_request(name, sentence))

print("\n【流程測試後的資料表】")
show_tables()
````
